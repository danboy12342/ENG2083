````python
# roguelike Room Objects - write your code here

#Coords x = collumns, y = rows
#forced into only for loops no if/while

#init funcs
def drawHorizontalWall(width):
    #draw a top or botom wall for a width given
    print("+", end="")
    for column in range(width):
        print("-", end="")
    print("+")

def drawEmptyWall(width):
    #draw a row of floor without the player
    print("|", end="")
    for column in range(width):
        print(".", end="")
    print("|")

def drawPlayerRow(width, playerX):
    #draw a row of floor with the player in it uisng @ for player
    print("|", end="")
    for column in range(playerX):#floor to left of player
        print(".", end="")
    print("@", end="")#player
    for column in range(width - playerX -1):#floor to right of player
        print(".", end="")
    print("|")

def drawRoom(width, height, playerX, playerY):
    #draws a room of given width and height with player at given coords
    drawHorizontalWall(width)
    for row in range(playerY):#rows above player
        drawEmptyWall(width)
    drawPlayerRow(width, playerX)#player
    for row in range(height - playerY - 1):#rows below player
        drawEmptyWall(width)
    drawHorizontalWall(width)
    print() #blank for interroom space

# init classes
class Room:
    def __init__(self, width, height): #take in width and heights
        self.width = width
        self.height = height
        self.playerX = 0 #init player coords as 00
        self.playerY = 0

#setup class funcs

    def draw(self): # draw player room
        drawRoom(self.width, self.height, self.playerX, self.playerY)

#left and right for drawing
    def left(self):
        if self.playerX > 0:
            self.playerX -= 1

    def right(self):
        if self.playerX < self.width - 1:
            self.playerX += 1

#and now up and down
    def up(self):
        if self.playerY > 0:
            self.playerY -= 1

    def down(self):
        if self.playerY < self.height - 1:
            self.playerY += 1



#main program

#call draw funcs
drawRoom(2,2,0,1)
drawRoom(5,3,1,2)
drawRoom(14,8,8,5)

#call class functions
myRoom = Room(4,4) #request room
myRoom.draw() #complete draw

#setup while loop to take in user input and move player around
while True:
    s = input() #get input
    if s=='a': myRoom.left() #left
    if s=='d': myRoom.right() #right
    if s=='w': myRoom.up() #you get the idea
    if s=='s': myRoom.down()
    if s=='x': #exit character
        print("all done")
        break
    myRoom.draw() # call draw after each input
````
