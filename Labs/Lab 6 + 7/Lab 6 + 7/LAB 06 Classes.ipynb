*(Oh the `IPython.display` library has a function called `clear_output` which will clear the old output so that you can overlay the new output and give the impression of movement)*

````python
# roguelike Room Objects - write your code here

#Coords x = collumns, y = rows
#forced into only for loops no if/while

#init class
class Room:
    def __init__(self, width, height): #pass in width and heights
        self.width = width
        self.height = height
        self.x = 0 #init coords at 00
        self.y = 0
        
    def draw(self):
        #draw top wall
        print("+", end="")
        for column in range(self.width):
            print("-", end="")
        print("+")

        #draw empty rows
        for row in range(self.y):
            print("|", end="")
            for column in range(self.width):
                print(".", end="")
            print("|")

        # drawout player row
        print("|", end="")
        for column in range(self.x): #left of plyr
            print(".", end="")
        print("@", end="")           # Player itself
        for column in range(self.width - self.x - 1): # right of player
            print(".", end="")
        print("|")

        #draw rows below
        for row in range(self.height - self.y - 1):
            print("|", end="")
            for column in range(self.width):
                print(".", end="")
            print("|")

        #draw bottom wall
        print("+", end="")
        for column in range(self.width):
            print("-", end="")
        print("+")

    # movment and collisoins
    def left(self):
        if self.x > 0: 
            self.x -= 1 #ensures player phase thro walls

    def right(self):
        if self.x < self.width - 1: 
            self.x += 1

    def up(self):
        if self.y > 0: 
            self.y -= 1

    def down(self):
        if self.y < self.height - 1: 
            self.y += 1

# Main program

#call class functions
myRoom = Room(4,4) #request room
myRoom.draw() #complete draw

#setup while loop to take in user input and move player around
while True:
    s = input() #get input
    if s=='a': myRoom.left() #left
    if s=='d': myRoom.right() #right
    if s=='w': myRoom.up() #you get the idea
    if s=='s': myRoom.down()
    if s=='x': #exit character
        print("all done")
        break
    myRoom.draw() # call draw after each input
````
