````python
# Write your set-up and calculations for question 1
'''
the idea here is pretty simple, we steal the code from the previous lab.
the outline is as follows:
'''
# var inits
launch_angle_deg = 25.0   # angles
v0 = 7.5                  # init v
g = 9.81                  # g std value

launch_angle = np.radians(launch_angle_deg) #conver to rad

plt.figure(figsize=[10, 10]) 

#acutal program

# init v components
vx0 = v0 * np.cos(launch_angle)   # hoz is cos
vy0 = v0 * np.sin(launch_angle)   # i dont have one for sin, its just the other one

# same as before see the reaosnsing in the last exercise
T = 2 * vy0 / g

# 500 steps for smoothness
time = np.linspace(0, T, 500)

# velocity
velocity_x = vx0 * np.ones_like(time)                   # no hoz , so const
velocity_y = vy0 - g * time                             # g
velocity_mag = np.sqrt(velocity_x**2 + velocity_y**2)   # v

# positions
position_x = vx0 * time                        # horz disp
position_y = vy0 * time - 0.5 * g * time**2    # alt

# disp out
print('Time in the air T = {0:.3g} s'.format(T))
print('Initial velocity: vx0 = {0:.3g} m/s, vy0 = {1:.3g} m/s'.format(vx0, vy0))
print('Minimum speed = {0:.3g} m/s'.format(velocity_mag.min()))
print('Maximum altitude = {0:.3g} m'.format(position_y.max()))
print('Landing distance = {0:.3g} m'.format(position_x[-1]))
````

````python
# Write your calculations for question 2 here

# init plot of velocity
plt.subplot(2, 2, 1)
plt.plot(time, velocity_x, label='velocity_x')
plt.plot(time, velocity_y, label='velocity_y')
plt.axhline(0, color='grey', linewidth=0.5)   # zero line: velocity_y crosses it at the top
plt.xlabel('Time / s')
plt.ylabel('Velocity / m s$^{-1}$')
plt.title('Horizontal and vertical velocity')
plt.legend()

#speeds
plt.subplot(2, 2, 2)
plt.plot(time, velocity_mag, color='purple')
plt.xlabel('Time / s')
plt.ylabel('Speed / m s$^{-1}$')
plt.title('Speed (velocity magnitude)')

# alt and horz disp
plt.subplot(2, 2, 3)
plt.plot(time, position_y, label='position_y (altitude)')
plt.plot(time, position_x, label='position_x (horizontal)')
plt.xlabel('Time / s')
plt.ylabel('Position / m')
plt.title('Altitude and horizontal displacement')
plt.legend()

#path
plt.subplot(2, 2, 4)
plt.plot(position_x, position_y, color='green')
plt.xlabel('Horizontal displacement / m')
plt.ylabel('Altitude / m')
plt.title('Projectile trajectory')

plt.tight_layout() #prevent overlapting
plt.savefig('projectile_motion.png')
plt.show()
````
