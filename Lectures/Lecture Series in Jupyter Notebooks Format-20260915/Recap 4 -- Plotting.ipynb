### Plot one-dimensional function $y = f(x)$
By using `plt.plot(x, y)` we can plot any graph of a continuous function such as $f(x) = x, f(x) = x^3$. For example, we can create a list of x-coorinates with `x = list(range(n))`. The list of y-coordinates can be generated by `for` loop and any kind of expression acting on `x`. One important problem with `range` is that it gives only list of integers.

````python
import matplotlib.pyplot as plt
import numpy as np
x = list(range(6))
y = []
for value in x:
    y.append(value * 3)
plt.plot(x, y)
plt.show() # comment this line out to see the return of plt.plot(x, y)
````

### Bar plot

### Plotting with data generated from `numpy` library

#### Why `numpy` library?

**Numpy** has powerful lists and arrays of doubles or floats. In fact, in practice `matplotlib.pyplot` is frequently used with `numpy` arrays instead of built-in Python lists. Some of arguments for using `numpy` library:
1. `numpy` arrays are designed for floating numbers, not like list for any data types. As for plots, we always need some sort of numeric values, not just any data types.
2. Operations on `numpy` arrays are built in C/C++ and executed extremely fast as compared to those computed on the built-in list. In fact, we need several `for` loops in dealing with lists while we don't need loops when dealing with `numpy` arrays most of the time.

***Further information for your own interest***

In fact, the `numpy` library is in C/C++ and designed to handle sophisticated floating operations (or numerical operations) on multi-dimensional arrays and multi-order tensors ([Wikipedia](https://en.wikipedia.org/wiki/Tensor)). It saves a lot of computational time because Python is a high-level programming language and it must do a lot of overheading for excuting for loop while C is a much lower-level programming and the C-code was compiled before being used.

**Note**  To understand the power of `numpy` library, we will revisit the library in the future lectures.

````python
# First of all, we need to import numpy
import numpy as np
# Let us import matplotlib.pyplot again here
import matplotlib.pyplot as plt
````

#### Using of `numpy.arange()` and `numpy.linspace()` for one-dimensional arrays

To define an array of evenly distanced grid points, we normally use the two following functions:
* `numpy.arange(<start>, <stop>, <step>)`
> generate an array spaning from `<start>` to `<stop>` with a predefined controlling step size `<step>`. ***Note***: The value `<stop>` is excluded from the array.

* `numpy.linspace(<start>, <stop>, <number-of-points>)`
> generate an array spanning from `<start>` to `<stop>` with a predefined number of grid points `<number-of-points>`. All grid points are evenly distanced from each other. ***Note***: The value `<stop>` is included in the array.

````python
# arange(<start>, <stop>, <step>)
x1 = np.arange(0.0, 10.0, 0.1)
# values from 0 to 10.0 in steps of 0.1, includes 0.0 and excludes 10.0
x1
# print(x1)  # Uncomment this statement to see the result in print()
````

````python
# linspace(<start>, <stop>, <number-of-points>)
x2 = np.linspace(0.0, 10.0, 15)
# values from 0 to 10.0 using 15 numbers equally distant from each other
# includes 0.0 and includes 10.0
x2
# print(x2)  # Uncomment this statement to see the result in print()
````

#### Examples of plotting using `numpy`
A huge benefit of ***numpy*** arrays is that **element-by-element** calculations don't need us to write an explicit **for** loop. Below are many more examples of using `matplotlib.plot`. These examples are pretty self-explantory. If not, the comments should do the job.

````python
x = np.linspace(0.0, 2*np.pi, 9)   # every 45 degree
y = np.sin(x)                       # we don't need a loop as we have had used built-in list in Python
plt.plot(x, y)
````

````python
# We can plot two functions in one common axis
x = np.linspace(0.0, 2*np.pi, 18)   # every 45 degree
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1)
plt.plot(x, y2)
plt.show() # just to remove the output of the last plt.plot
````

````python
# We can make the plot looks more interesting by using "color", "marker" and "label"
x = np.linspace(0.0, 2*np.pi, 25)
y = np.sin(x)
# b --> blue, o --> circle marker, dash [-] --> continous line
plt.plot(x, y, 'bo-', label='sin(x)')   # plot with style and label (legend)
# loc --> keyword standing for location
plt.legend(loc='upper right')           # location of the legend
plt.xlabel('x')                         # label for x-axis
plt.ylabel('y')                         # label for y-axis
# plt.show()                            # uncomment this line, then the output of plt.ylabel() disappears
````

````python
# Scatter plots
x = np.linspace(0.0, 2*np.pi, 20)
# c --> color, 'g' --> green, 'marker' is self-explanatory
plt.scatter(x, np.sin(x), c='g', marker='x')    # just use y = np.sin(x) directly
# 'm' --> magenta, 's' --> square
plt.scatter(x, np.cos(x), c='m', marker='s')    # just use y = np.cos(x) directly
plt.show()                                      # once again: to remove the output of the last plot command
````

We can use also color map in Python. Color map is a rather advanced concept. However, in simple words, we can think of it as a mapping from value to a set of colors that will be used for representation. The color map can be defined by using the keyword argument `cmap=something`. Some of more useful functions for genearting eye-pleasing plot.
- `plt.title("title string")`
- `plt.tick_params()`
- `plt.axis([xmin, xmax, ymin, ymax])`

````python
# cmap --> colormap
x = np.linspace(0.0, 2*np.pi, 25)
y = np.sin(x)
plt.scatter(x, y, c=x, cmap=plt.cm.Blues, marker='x')
plt.title("Sine wave", fontsize=18)     # of course title of the plot
plt.tick_params(axis='both', which='major', labelsize=12)   # ticks of axis
# xmin = -0.5, xmax = 10, ymin = -1, ymax = 1
plt.axis([-0.5, 10, -1, 1]) 

# and over-plotting also works
plt.scatter(x[0], y[0], c='g', s=100)       # s --> size of the marker
plt.scatter(x[1], y[1], c='r', s=200)       # s --> size of the marker
plt.show()                                  # to remove the output of the last plot command
````

````python
# turn off axes, make the plot as big or small as we like
axes = plt.axes()
axes.get_xaxis().set_visible(False)
````

````python
# Set figure size (figsize) and number of dots per inch (dpi)
plt.figure(dpi=128, figsize=(2.5, 1.5)) 
# thinner line, smaller markers
plt.plot(x, np.sin(x), 'k^-', markersize=3, linewidth=0.5)
# thicker line, bigger markers
plt.plot(x, np.cos(x), 'rv-', markersize=5, linewidth=1.0)  
plt.title("Sine wave", fontsize=8)

# After all the plotting, we can also save our plot to file
plt.savefig("sine_function.png", bbox_inches='tight')
````

# Files

- Working with files is generally quite tricky.
- Writing to text files seems easy; you just squirt data out and whoever picks up the file has to deal with the formatting in the file.
- What happens if something goes wrong with the disk half way through the write?
- Reading files is trickier since we will need to know enough information about comprehension of formatting to turn the raw text into useful data.

````python
file_name = 'my_file.txt'
f = open(file_name, 'w', encoding='utf-8') # the default of encoding is platform dependent
# We should see the file with name "my_text.txt" after this command.

# We should/must close the file in the end
f.close()
````

It is good practice to use the `with` keyword when dealing with file objects. The advantage is that the file is properly closed after its suite finishes, even if an exception is raised at some point. Using `with` is also much shorter than writing equivalent `try-finally` blocks.

````python
file_name = 'my_file.txt'
with open(file_name, 'w') as f:     # do it this way!
    f.write('a first line of text\n')
    f.write('a second line of text\n')
    f.write('yeah finally third line.\n')
# \n ensures end of line characters are put into the text file properly

# After this block of statements, you can click on the "my_file.txt" to see the content. It should appear
'''
a first line of text
a second line of text
yeah finally third line.
'''
````

Note that we have used the writing mode with the signature `'w'` in the second argument of `open(file_name, 'w')`. The following table summraizes multiple modes when working with files.

Mode | Description
-----|--------------
r    | open a file for reading (default)
w    | open a file for writing. Creates a new file if it does not exist or truncates the file if it exists
x    | open a file for exclusive creation. If the file already exists, the operation fails
a    | open a file for appending at the end of the file without truncating it, create a new file if it does not exist
t    | open in text mode (default)
b    | open in binary mode
\+   | opens a file for updating (reading and writing) 

In the following, we do some more examples using the modes in the table.

````python
# Appending text to the text file
file_name = 'my_file.txt'  # declare just for self-contained block of code
with open(file_name, 'a') as f:  # open the file for appending
    f.write('a fourth line of text\n')
    f.write('a fifth line of text\n')
    f.write("obviously the 'a' means append\n" )
````

### Methods of File Objects

To read a file's contents, call `f.read(size)`, which reads some quantity of data and returns it as a string (in text mode) or bytes object (in binary mode). `size` is an optional numerica argument. When `size` is omitted or negative, the entire contents of the file will be read and returned. If the end of the file has been reached, `f.read()` will return empty string (`''`).

`f.readline()` reads a single line from a file; a newline character (`\n`) is left at the end of the string, and is only ommited on the last line of the file if the file does not end in a newline.

`f.write(string)` writes the contents of `string` to the file, returning the number of characters written.

`f.tell()` returns an integer giving the file object's current position in file represented as number of bytes from the beginning of the file when in binary mode and an opaque number when in text mode.

`f.seek(offset, whence)` changes the file object's position. The position is computed from adding `offset` to a reference point. The reference point is selected by the `whence` argument. A `whence` value of $0$ measures from the beginning of the file, $1$ uses the current file position, and $2$ uses the end of the file as the reference point, `whence` can be omitted and defaults to $0$, using the beginning of the file as the reference point.

````python
with open(file_name, 'r') as f:
    full_content = f.read()

print(full_content)
````

````python
with open(file_name, 'r') as f:
    line1 = f.readline()
    line2 = f.readline()
    line3 = f.readline()
print(line1)
print(line2)
print(line3)
````

For reading lines from a file, we can loop over the file object. This is memory efficient, fast and leads to simple code. Why is it memory efficient? We don't store the entire file into the variable containing the file content. We only store one line by each `f.readline()`.

````python
# read line by line
with open(file_name, 'r') as f:
    for line in f:
        print(line.rstrip())
        
# rstrip() gets rid of extra end-of-line characters added by python. Compare the output of this block with the output of the above block.
````

````python
# We can store the lines in the list of strings
with open(file_name) as f:
    lines = f.readlines()
    
for line in lines:
    print(line.rstrip())
````
